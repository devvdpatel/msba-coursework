# Task 3: 
Given a list of integers, write a program to find the three largest numbers in the list without using built-in sorting functions. 
Return the numbers in a list. If there are fewer than three distinct numbers in the list, return a message stating 'Not enough distinct numbers'. 

```python
#Task 3: Given a list of integers, write a program to find the three largest numbers in the list without using built-in sorting functions. 
# Return the numbers in a list. If there are fewer than three distinct numbers in the list, return a message stating 'Not enough distinct numbers'. 
def task3_function(input_list):
    
    #get rid of duplicates
    new_list = list(set(input_list))
    #check if the length of the new list is less than 3
    if len(new_list) < 3:
        return "Not enough distinct numbers"
    
    #variables to store values
    first, second, third = 0, 0, 0
    
    #iterate through new list
    for i in new_list:
        if i > first:
            third = second
            second = first
            first = i
        elif i > second:
            third = second
            second = i
        elif i > third:
            third = i
    return [first, second, third]


#Test Case 1
list1 = [5, 23, 8, 23, 12, 5, 8]
print(task3_function(list1))

#Test Case 2
list2 = [7, 5, 5]
print(task3_function(list2))
```

# Task 4
Sum all values in the list objects only if they are of type tuple

```python
list1 = [[1,2,3], (4,5,6), 2, "2"]

def find_sum(input_list):
    tuple_sum = 0
    #iterate through the list
    for i in input_list:
        if isinstance(i, tuple):
            tuple_sum += sum(i)
    return tuple_sum

print(find_sum(list1))
    
```

# Task 7 
Combine the values of common keys by multiplying them 

Suppose 
d1 = {'a': 100, 'b': 200, 'c':300}
d2 = {'a': 3, 'b': 2, 'd':4}
The expected output should be
{'a': 300, 'b': 400,'c': 300, 'd': 4,}


```python
def task7(dict1, dict2):
    result = {}
    
    #iterate through the union of the two dictionaries
    for key in dict1 | dict2:
        #condition1: key is in both dictionaries
        if key in dict1 and key in dict2:
            result[key] = dict1[key] * dict2[key]
        #condition2: key is only in dict1
        elif key in dict1:
            result[key] = dict1[key]
        #condition3: key is only in dict2
        else:
            result[key] = dict2[key]
    
    return(result)

d1 = {'a': 100, 'b': 200, 'c':300}
d2 = {'a': 3, 'b': 2, 'd':4}

task7(d1, d2)
```

# Task 8 
We store cargo in some location identified with 3 different numbers (x, y, and height) in the format of a tuple. For the example, the cargo variable looks like this:

Cargo={(30,23,12): {"a":12,"b":13,"c":15},(36,3,122):{"b":15,"d":3,"e":5}, (36,3,12):{"a":25,"c":30,"f":5}}

It says that at location (30,23,12), there are 12 “a” cargo, 13 “b” cargo, and 15 “c” cargo. Note that the value of the outer dictionary is itself a dictionary.

We also have a price chart stored in the format of a dictionary, showing the price of different types of cargo. For example: 

price={"a":100,"b":200,"c":300,"d":450,"e":700,"f":650}
Which says that the price of “a” is $100.

Calculate the total value of cargo stored at the x-coordinate of 36


```python
cargo = {(30,23,12): {"a":12,"b":13,"c":15},(36,3,122):{"b":15,"d":3,"e":5}, (36,3,12):{"a":25,"c":30,"f":5}}
price = {"a":100,"b":200,"c":300,"d":450,"e":700,"f":650}

x_coord = 36

def calculate_total(c, p, x):
    #initialize variable to store the total value
    total_value = 0
    #iterate through the cargo dictionary using .items()
    for location, items in c.items():
        #check if x-coordinate of the location is equal to the given x
        if location[0] == x:
            #iterate through the items in the location
            for item, quantity in items.items():
                #multiply the quantity by the price and add it to the total value
                total_value += quantity * p[item]
    return total_value

total_value = calculate_total(cargo, price, x_coord)

print(f"The total value of cargo stored at the x_coord of {x_coord} is {total_value}")
```

# Task 9:
Please write code to identify the last recurring character in a string, its positions, and the number of times it recurs. If no recurring characters are found, return a message stating so. Also, if there are multiple characters that recur for the maximum number of times, return all of them with their positions (input = "XDZYTUVZWV", "XDZZTUVZWT", "XDYUWV")
For example:
•	input = "XDZYTUVZWV", output = {'Z': [2, 7], 'V': [6, 9], "times_recurring": 2}
•	input = "XDZZTUVZWT", output = {"Z": [2, 3, 7], "T": [4, 9], "times_recurring": 3, 2}
•	input = "XDYUWV", output = "No recurring characters found"


```python
def task9(s):
    #empty dict
    character_positions = {}
    #iterate over the string
    for i, char in enumerate(s):
        #if the character is in the dictionary, append the index to the list
        if char in character_positions:
            character_positions[char].append(i)
        #else add the character to the dictionary
        else:
            character_positions[char] = [i]

    #empy dict to store recurring characters
    recurring_chars = {}

    #count of max recurring characters
    count = 0
    #iterate over the dictionary
    for char, positions in character_positions.items():
        if len(positions) > 1:
            recurring_chars[char] = positions
            count = max(count, len(positions))

    if not recurring_chars:
        return "No recurring characters found"

    result = {}
    for char, positions in recurring_chars.items():
        if len(positions) == count:
            result[char] = positions
    result["times_recurring"] = count
    return result

# Test cases
test1 = ["XDZYTUVZWV"]
test2 = ["XDZZTUVZWT"]
test3 = ["XDYUWV"]


print(task9(test1[0]))
print(task9(test2[0]))
print(task9(test3[0]))


    
```

# Task 10
Design a function that takes in two strings ('what', 'where'). For each string, it should remove characters that appear in the other string. After the removal, it should concatenate the first 50% of the modified first string with the last 50% of the modified second string. If either of the modified strings has less than 2 characters, return a message indicating which string is too short.

For example:
string_1 = "programming", string_2 = "language", the function should return `prorue`.
string_1 = "hello", string_2 = "world", the function should return `hrd`.
string_1 = "a", string_2 = "b", the function should return "String 1 is too short after modifications."
Hint: Consider using string slicing for effective concatenation.


```python
def task10(string_1, string_2):
    
    # new string with characters from string_1 that are not in string_2
    new1 = ''.join([char for char in string_1 if char not in string_2])
    
    # new string with characters from string_2 that are not in string_1
    new2 = ''.join([char for char in string_2 if char not in string_1])
    
    # Initial condition check
    if len(new1) < 2:
        return "String 1 is too short."
    elif len(new2) < 2:
        return "String 2 is too short."
    
    # Get first half of new1 string
    half_len_1 = len(new1) // 2
    first_half_1 = new1[:half_len_1]
    
    # Get the second half from new2 string
    half_len_2 = len(new2) // 2
    last_half_2 = new2[half_len_2:]
    
    # Concatenate the two halves and return the result
    return first_half_1 + last_half_2

# Test cases
print(task10("programming", "language"))  
print(task10("hello", "world"))           
print(task10("a", "b"))                  
```
