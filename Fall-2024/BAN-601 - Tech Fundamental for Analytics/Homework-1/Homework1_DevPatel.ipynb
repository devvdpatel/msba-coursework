```python
import pandas as pd

data3 = {'ID': [1, 2, 3, 4], 'Score': [85, 90, 88, 78]} 
print(data3['ID'][2], data3['Score'][2])

data4 = {'ID': [3, 4, 5], 'Bonus': [5, 3, 2]} 

df3 = pd.DataFrame(data3)
df4 = pd.DataFrame(data4)

right_join = pd.merge(df3, df4, on='ID', how='right')
print(right_join)

left_join = pd.merge(df3, df4, on='ID', how='left')
print(left_join)
```

```python
def task7(any_string):
    if len(any_string) < 3:
        return False
    
    #extract last three characters
    last_three = any_string[-3:]
    
    for i in range(len(any_string) - 3):
        if any_string[i:i+3] == last_three:
            return True
    return False

test_string = "DevDevDevDev"
print(task7(new_string))
print(task7(test_string))
```
